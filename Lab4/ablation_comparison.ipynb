```python
# ============================================================
# RESULTS FROM ALL EXPERIMENTS
# ============================================================
# Fill in these values from each experiment's summary.txt file

results = {
    'baseline': {
        'train_loss': 0.3493,
        'val_loss': 0.6931,
        'train_val_gap': 0.3438,
        'time_hours': 7.75,
        'best_epoch': 9,
        'modifications': 'None (original baseline)',
        'notes': 'Clear overfitting - large train-val gap'
    },
    'mod1_dropout': {
        'train_loss': 0.0,  # TODO: Fill in after running experiment
        'val_loss': 0.0,     # TODO: Fill in after running experiment
        'train_val_gap': 0.0,
        'time_hours': 0.0,
        'best_epoch': 0,
        'modifications': 'Dropout (0.3) + Label Smoothing (0.1)',
        'notes': 'Trained from scratch to integrate regularization from start'
    },
    'mod2_augment': {
        'train_loss': 0.0,  # TODO: Fill in after running experiment
        'val_loss': 0.0,     # TODO: Fill in after running experiment
        'train_val_gap': 0.0,
        'time_hours': 0.0,
        'best_epoch': 0,
        'modifications': 'Strong Data Augmentation (CLIP-aware)',
        'notes': 'Fine-tuned from baseline checkpoint'
    },
    'mod3_combined': {
        'train_loss': 0.0,  # TODO: Fill in after running experiment
        'val_loss': 0.0,     # TODO: Fill in after running experiment
        'train_val_gap': 0.0,
        'time_hours': 0.0,
        'best_epoch': 0,
        'modifications': 'All modifications (Dropout + Label Smoothing + Augmentation)',
        'notes': 'Trained from scratch with all regularization techniques'
    },
}

print("Results loaded successfully!")
print(f"\nExperiments to compare: {len(results)}")
for exp_name in results.keys():
    print(f"  - {exp_name}")
```

```python
# ============================================================
# GENERATE COMPARISON TABLE AND ANALYSIS
# ============================================================

import pandas as pd

# Create comparison dataframe
comparison_df = pd.DataFrame({
    'Experiment': list(results.keys()),
    'Train Loss': [results[exp]['train_loss'] for exp in results.keys()],
    'Val Loss': [results[exp]['val_loss'] for exp in results.keys()],
    'Train-Val Gap': [results[exp]['train_val_gap'] for exp in results.keys()],
    'Time (hours)': [results[exp]['time_hours'] for exp in results.keys()],
    'Best Epoch': [results[exp]['best_epoch'] for exp in results.keys()],
})

print("=" * 80)
print("ABLATION STUDY COMPARISON TABLE")
print("=" * 80)
print(comparison_df.to_string(index=False))
print("=" * 80)

# Find best performing experiment
best_val_exp = min(results.keys(), key=lambda x: results[x]['val_loss'])
best_gap_exp = min(results.keys(), key=lambda x: results[x]['train_val_gap'])

print("\nKEY FINDINGS:")
print("-" * 80)
print(f"Best validation loss: {best_val_exp} ({results[best_val_exp]['val_loss']:.4f})")
print(f"Best generalization: {best_gap_exp} (train-val gap: {results[best_gap_exp]['train_val_gap']:.4f})")
print("-" * 80)
```

```python
# ============================================================
# ANALYSIS: Impact of Each Modification
# ============================================================

print("=" * 80)
print("IMPACT ANALYSIS")
print("=" * 80)

baseline_val = results['baseline']['val_loss']
baseline_gap = results['baseline']['train_val_gap']

print("\n1. Modification 1 (Dropout + Label Smoothing):")
print("-" * 80)
mod1_val_improvement = baseline_val - results['mod1_dropout']['val_loss']
mod1_gap_improvement = baseline_gap - results['mod1_dropout']['train_val_gap']
print(f"   Validation loss change: {mod1_val_improvement:+.4f}")
print(f"   Train-val gap change: {mod1_gap_improvement:+.4f}")
print(f"   Impact: {'Positive' if mod1_val_improvement > 0 else 'Negative' if mod1_val_improvement < 0 else 'Neutral'}")

print("\n2. Modification 2 (Strong Augmentation):")
print("-" * 80)
mod2_val_improvement = baseline_val - results['mod2_augment']['val_loss']
mod2_gap_improvement = baseline_gap - results['mod2_augment']['train_val_gap']
print(f"   Validation loss change: {mod2_val_improvement:+.4f}")
print(f"   Train-val gap change: {mod2_gap_improvement:+.4f}")
print(f"   Impact: {'Positive' if mod2_val_improvement > 0 else 'Negative' if mod2_val_improvement < 0 else 'Neutral'}")

print("\n3. Modification 3 (Combined):")
print("-" * 80)
mod3_val_improvement = baseline_val - results['mod3_combined']['val_loss']
mod3_gap_improvement = baseline_gap - results['mod3_combined']['train_val_gap']
print(f"   Validation loss change: {mod3_val_improvement:+.4f}")
print(f"   Train-val gap change: {mod3_gap_improvement:+.4f}")
print(f"   Impact: {'Positive' if mod3_val_improvement > 0 else 'Negative' if mod3_val_improvement < 0 else 'Neutral'}")

print("\n" + "=" * 80)
```
